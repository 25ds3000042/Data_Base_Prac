```python
import pandas as pd
import sqlite3

print("Libraries imported successfully!")
```

```python
students = {
    "student_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "name": [
        "Rahul", "Priya", "Amit", "Sneha",
        "Karan", "Anjali", "Vikram", "Neha"
    ],
    "age": [22, 23, 21, 24, 22, 25, 23, 21],
    "city": [
        "Bangalore", "Mumbai", "Delhi", "Bangalore",
        "Pune", "Delhi", "Mumbai", "Pune"
    ],
    "marks": [85, 92, 76, 88, 67, 95, 72, 89]
}

df_students = pd.DataFrame(students)

df_students
```

```python
courses = {
    "course_id": [101, 102, 103, 104],
    "course_name": [
        "Python",
        "Machine Learning",
        "SQL",
        "Deep Learning"
    ],
    "duration_months": [3, 4, 2, 5]
}

df_courses = pd.DataFrame(courses)

df_courses
```

```python
enrollments = {
    "enrollment_id": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "student_id": [1, 1, 2, 3, 4, 5, 6, 7, 8],
    "course_id": [101, 103, 102, 101, 104, 103, 102, 104, 101],
    "fee": [5000, 3000, 8000, 5000, 10000, 3000, 8000, 10000, 5000]
}

df_enrollments = pd.DataFrame(enrollments)

df_enrollments
```

```python
df_students.to_csv("students.csv", index=False)
df_courses.to_csv("courses.csv", index=False)
df_enrollments.to_csv("enrollments.csv", index=False)

print("CSV files created!")
```

Establish database connection

```python
conn = sqlite3.connect("student_management.db")

cursor = conn.cursor()

print("Database created and connected!")
```

```python
cursor.execute("""
CREATE TABLE students (
    student_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    city TEXT,
    marks REAL
)
""")

conn.commit()

print("Students table created!")
```
